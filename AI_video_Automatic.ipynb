```python
import whisper
import os

print("Đang tải mô hình AI Whisper bản 'base' (vừa nhanh vừa chính xác cho tiếng Việt)...")
model = whisper.load_model("base")
print("Đã tải xong mô hình AI! Sẵn sàng xử lý video bất động sản.")
```

```python
from google.colab import files
import os

print("Hãy chọn file video thô trên máy tính của anh để tải lên Google Colab:")
uploaded = files.upload()

# Lấy tên file video vừa tải lên
for filename in uploaded.keys():
    video_file = filename
    print(f"\nĐã tải lên thành công: {video_file}")

    print("AI Whisper đang bắt đầu nghe audio và tạo phụ đề tiếng Việt (quá trình này mất từ 1-3 phút tùy độ dài video)...")
    result = model.transcribe(video_file, language="vi")

    print("\n--- KẾT QUẢ PHỤ ĐỀ (SUB) AI VỪA TẠO ---")
    # In ra 5 dòng đầu tiên của phụ đề để kiểm tra
    for segment in result["segments"][:5]:
        print(f"[{segment['start']:.1f}s --> {segment['end']:.1f}s]: {segment['text']}")
    print("...")
    print("AI đã dịch và tạo phụ đề thành công toàn bộ video!")
```

```python
import os

# Hàm tạo file phụ đề chuẩn .srt từ kết quả của AI
srt_filename = "subtitles.srt"
with open(srt_filename, "w", encoding="utf-8") as f:
    for i, segment in enumerate(result["segments"], start=1):
        start_time = segment["start"]
        end_time = segment["end"]
        text = segment["text"].strip()

        # Chuyển đổi giây sang định dạng giờ:phút:giây,mili-giây của phụ đề
        def format_time(seconds):
            hours = int(seconds // 3600)
            minutes = int((seconds % 3600) // 60)
            secs = int(seconds % 60)
            millisecs = int((seconds - int(seconds)) * 1000)
            return f"{hours:02d}:{minutes:02d}:{secs:02d},{millisecs:03d}"

        f.write(f"{i}\n{format_time(start_time)} --> {format_time(end_time)}\n{text}\n\n")

print(f"Đã tạo file phụ đề chuẩn: {srt_filename}")

# Tên video đầu ra hoàn chỉnh
output_video = "video_hoan_chinh.mp4"

# Dùng FFmpeg để nhúng phụ đề vào video gốc (đính kèm font chữ và kiểu chữ rõ nét)
print("Đang tiến hành ghép phụ đề và xuất video hoàn chỉnh...")
# Lệnh ffmpeg nhúng sub trực tiếp lên video
os.system(f"ffmpeg -y -i {video_file} -vf \"subtitles={srt_filename}:force_style='FontSize=24,PrimaryColour=&H00FFFF&,Outline=2,Shadow=1'\" -c:a copy {output_video}")

print(f"\n🎉 HOÀN TẤT! Video chuyên nghiệp đã sẵn sàng: {output_video}")
```
